```
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

df = sns.load_dataset('titanic')
df.to_csv('titanic.csv', index=False)
print(df.info())
print(df.describe())
print(df.isnull().mean()*100)
plt.hist(df['age'].dropna(), bins=20)
plt.title('Age Distribution')
plt.show()
```
